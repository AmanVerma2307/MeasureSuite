```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
labels = np.load('./Embeddings/y_dev_sensor_Acc_seqLen150_bdb.npz')['arr_0']
labelsId =  np.load('./Embeddings/y_dev_id_sensor_Acc_seqLen150_bdb.npz')['arr_0']
```

```python
dataMat_id = np.zeros(shape=(51,1))
dataMat_gest = np.zeros(shape=(4,1))
dataMat = np.zeros(shape=(4,51))

def plotGramMatrix(cm,
                   cmap=plt.cm.Blues):
    """
    This function prints and plots the confusion matrix.
    Normalization can be applied by setting `normalize=True`.
    """
    ax = sns.heatmap(cm, cmap=cmap, linewidth=0.5, linecolor='black')
    plt.show()
```

```python
for idx, labelCurr in enumerate(labels):
    dataMat_gest[labelCurr,:] = dataMat_gest[labelCurr,:] + 1

plt.bar(np.arange(1,5),dataMat_gest[:,0])
plt.show()
```

```python
for idx, labelCurr in enumerate(labelsId):
    dataMat_id[labelCurr,:] = dataMat_id[labelCurr,:] + 1

plt.bar(np.arange(1,52),dataMat_id[:,0])
plt.show()
```

```python
for labelGestCurr, labelIdCurr in zip(labels,labelsId):
    dataMat[labelGestCurr,labelIdCurr] = dataMat[labelGestCurr,labelIdCurr] + 1

print(dataMat)
plotGramMatrix(dataMat)
```

```python
plotGramMatrix(dataMat > 1)
```
